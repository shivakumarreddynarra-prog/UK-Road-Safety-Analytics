```python
def missing_report(df):
    report = pd.DataFrame({
        "missing_count": df.isnull().sum(),
        "missing_percent": (df.isnull().sum() / len(df) * 100).round(2)
    })

    return report[report["missing_count"] > 0].sort_values(
        "missing_percent",
        ascending=False
    )

print("COLLISIONS - MISSING VALUES")
display(missing_report(collisions))

print("\nVEHICLES - MISSING VALUES")
display(missing_report(vehicles))

print("\nCASUALTIES - MISSING VALUES")
display(missing_report(casualties))
```

```python
datasets = {
    "Collisions": collisions,
    "Vehicles": vehicles,
    "Casualties": casualties
}

for name, df in datasets.items():
    print(f"\n{name.upper()}")

    for dtype in ["str", "object", "float64"]:
        columns = df.select_dtypes(include=[dtype]).columns.tolist()
        print(f"{dtype}: {columns}")
```

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    
    non_numeric = df.select_dtypes(exclude=["number"]).columns.tolist()
    float_columns = df.select_dtypes(include=["float64"]).columns.tolist()
    
    print("Non-numeric columns:", non_numeric)
    print("Float columns:", float_columns)
```
